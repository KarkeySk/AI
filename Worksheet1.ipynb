```python
#Problem 4
#1
#calculates the inverse of matrix and proves that A.A(inverse)= I
A = np.array([[3, 4], [7, 8]])

aInv = np.linalg.inv(A)
result = np.dot(A,aInv)
print("A inv is :" , aInv)
print("A * Ainv: ",result )
```

```python
#2
#multiplies matrix A and B in order AB and BA to prove AB!=BA
A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])
AB = np.dot(A,B)
BA = np.dot(B,A)
print(AB)
print(BA)
```

```python
#3
#Verifies that (AB)T = BT AT , T = Transverse
A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

AB = np.dot(A,B)

left = AB.T

aFlip = A.T
bFlip = B.T

right  = np.dot(bFlip,aFlip)
print(left)

print(right)
```

```python

#4
#Solves a system of linear equation Ax = B
A = np.array([
    [ 2, -3,  1],
    [ 1, -1,  2],
    [ 3,  1, -1]
])

B = np.array([-1, -3, 9])

solution = np.linalg.solve(A, B)

print("Solution", solution)
```

```python
#4.2
#Comparing addition speed between python list and numpy array.
import time
import random
import numpy as np

list1 = [random.random() for _ in range(1_000_000)]
list2 = [random.random() for _ in range(1_000_000)]

#Python list addition
start = time.time()
list_add = [list1[i] + list2[i] for i in range(1_000_000)]
end = time.time()
print("Python list addition time:", end - start, "seconds")
arr1 = np.random.rand(1_000_000)
arr2 = np.random.rand(1_000_000)

#Numpy array addition
start = time.time()
np_add = arr1 + arr2
end = time.time()
print("NumPy addition time:", end - start, "seconds")

```

```python
#Comparing multication speed between python list and numpy array.

#python
start = time.time()
list_mul = [list1[i] * list2[i] for i in range(1_000_000)]
end = time.time()
print("Python list multiplication time:", end - start, "seconds")

#numpy
start = time.time()
np_mul = arr1 * arr2
end = time.time()
print("NumPy multiplication time:", end - start, "seconds")
```
