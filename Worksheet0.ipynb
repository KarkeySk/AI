```python
# Daily time (in hours): [study, entertainment, sleep]

time_data = [
    (3.5, 2.0, 7.0), (5.0, 1.5, 6.5), (2.5, 3.0, 8.0),
    (4.0, 2.0, 6.0), (1.5, 4.5, 9.0), (3.0, 2.5, 7.5),
    (5.5, 1.0, 6.0), (2.0, 3.5, 8.5), (4.5, 2.0, 7.0),
    (3.0, 3.0, 7.5), (6.0, 1.5, 6.0), (2.5, 4.0, 8.0),
    (4.0, 2.5, 7.0), (5.0, 2.0, 6.5), (3.5, 2.5, 7.0)
]

#Task 1:
#Classify Study Time:
"""
The below code categorizes the studyHrs into High, Low and Moderate
"""
Low=[]
Moderate =[]
High =[]

for day in time_data: #Day will become a tuple and then studyhrs can borrow day ko first value while the loop procceds
    studyHrs= day[0]
    if studyHrs <3:
        Low.append(studyHrs)
    elif studyHrs>=3 and studyHrs<=5:
        Moderate.append(studyHrs)
    else: High.append(studyHrs)
print(f"The Low study times were {Low}")
print(f"The Moderate study times were {Moderate} ")
print(f"The High study times were {High}")
```

```python
#Task2
#Based on Data:
"""
The below code provides the total times that were studied for lowest,moderate, highes.
"""
#1
lowCount= len(Low)
print(f"The lowest studied {lowCount} times")

#2
moderateCount = len(Moderate)
print(f"The Moderate studied {moderateCount} times")

#3
highCount = len(High)
print(f"The Highest studied {highCount} times")
```

```python
#Task 3
#Convert Study Hours to Minutes:

studyMin =[]
comboStudyHrs = Low + Moderate + High
for hrs in comboStudyHrs:
    studyMin.append(hrs*60)
print(studyMin)
```

```python
#Task 4
#Analyze Average Time Use:

"""
The below code calculates and prints the average of study,entertainment and sleep hrs.
"""
newStudyHrs = []
entertainmentHrs =[]
sleepHrs = []

for day in time_data:
    s, e , sl = day
    newStudyHrs.append(s)
    entertainmentHrs.append(e)
    sleepHrs.append(sl)
    avgStudy= sum(newStudyHrs)/len(newStudyHrs)
    avgEnt= sum(entertainmentHrs)/len(entertainmentHrs)
    avgsleephrs= sum(sleepHrs)/len(sleepHrs)
print("Average study hr is",avgStudy)
print("Average enter hr is",avgEnt)
print("Average sleep hr is",avgsleephrs)
```

```python
#Recursion:
#Task 1

def sumNestedList(nested_list):
  """
    This provides the sum of all nested list provided.

    Arguments: nested_list

    Returns: The sum of all the nested list.

  """
  total = 0

  for element in nested_list:
      if isinstance(element,list):
          total+=sumNestedList(element)
      else:
          total += element
  return total

nested_list = [1, [2, [3, 4], 5], 6, [7, 8]]
print(sumNestedList(nested_list))
```

```python
#Task 2

def generate_permutations(s):
  """
  This generates all permutations of strings.
  Arguments: s: string
  Returns: Will provide all possible permutations.

  """
  if len(s) <= 1:
      return [s]

  permutations = []

  for i in range(len(s)):
      char = s[i]

      remaining = s[:i] + s[i+1:]

      for perm in generate_permutations(remaining):
          permutations.append(char + perm)

  return permutations
print(generate_permutations("abc"))
```

```python
#Task 3

directory_structure = {
    "file1.txt": 200,
    "file2.txt": 300,
    "subdir1": {
        "file3.txt": 400,
        "file4.txt": 100
    },
    "subdir2": {
        "subsubdir1": {
            "file5.txt": 250
        },
        "file6.txt": 150
    }
}

def directorSizer(directory):
  """
  This calcualtes all size of files in a nested directory.
  Arguments: dictionary: A dict with subdirectories.
  returns: Total size of all files.
  """
  totalSize = 0

  for item in directory:
      value = directory[item]

      if isinstance(value,dict):
          totalSize += directorSizer(value)
      else:
          totalSize +=value

  return totalSize

print(directorSizer(directory_structure))
```

```python
#Dynamic Programming:
#Task 1

def min_coins(coins, amount):
    """
    It calculates the minimum number of coins that is needed to reach the amount required with the coins that are provided.

    Args: coins:list of integers
          amount: the total amount needed
    Returns: The minimum number of coins needed to reach that amount. -1 if the amount cannot be reached.
    """
    if amount == 0:
        return 0

    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    for coin in coins:
        for i in range(coin, amount + 1):
            dp[i] = min(dp[i], dp[i - coin] + 1)

    return dp[amount] if dp[amount] != float('inf') else -1
print(min_coins([1, 2, 5], 11))

```

```python
#Task 2
def longest_common_subsequence(s1, s2):
    """
    Calculates the longest length of a common subsequence between two strings

    Args: s1 : First string
          s2 : Second string.

    Returns: The length of the longest common subsequence of the s1 and s2 strings.
    """
    n = len(s1)
    m = len(s2)

    dp = [[0] * (m + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = 1 + dp[i - 1][j - 1]
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    return dp[n][m]


print(longest_common_subsequence("abcde", "ace"))


```

```python
#Task 3
def knapsack(weights, values, capacity):
    """
    It picks items so that the total weight doesn't go over the capacity.

    Args: weights: Weight of item.
          values: Value of item.
          capacity: Max weight the knapsack can withstand.

    Returns: Highest total value that is possible.

    """
    n = len(weights)

    dp = []
    for i in range(n + 1):
        row = []
        for j in range(capacity + 1):
            row.append(0)
        dp.append(row)

    for i in range(1, n + 1):
        for current_capacity in range(1, capacity + 1):
            item_weight = weights[i - 1]
            item_value = values[i - 1]

            dont_take = dp[i - 1][current_capacity]

            if item_weight <= current_capacity:
                take = item_value + dp[i - 1][current_capacity - item_weight]
                dp[i][current_capacity] = max(take, dont_take)
            else:
                dp[i][current_capacity] = dont_take

    return dp[n][capacity]

weights = [1, 3, 4, 5]
values = [1, 4, 5, 7]
capacity = 7

print(knapsack(weights, values, capacity))
```
