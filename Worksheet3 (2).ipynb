##Exploratory Data Analysis with Group-by Method Practice Exercise:

###percent of the total fare revenue came from each passenger

```python
#1
sum_Fare = df['Fare'].sum()
sum_Fare
```

```python
#2
fare_class1 = df[df['Pclass']== 1]['Fare'].sum()
print("First class total fare",fare_class1)
```

```python
#3
fare_class2 = df[df['Pclass']== 2]['Fare'].sum()
print("Second class total fare",fare_class2)


fare_class3 = df[df['Pclass']== 3]['Fare'].sum()
print("Third class total fare",fare_class3)
```

```python
#4
fare_total = [fare_class1,fare_class2,fare_class3]
print("The total fare is list:",fare_total)
```

```python
#5
proportions =[]
for fare in fare_total:
  percentage = (fare/ sum_Fare) *100
  proportions.append(percentage)
print(proportions)
```

###percent of the total number of passengers on the Titanic belonged to each age group

```python
#1
df.loc[df['Age'] < 18, 'age_group'] = 'child'
df.loc[(df['Age'] >= 18) & (df['Age'] < 65), 'age_group'] = 'adult'
df.loc[df['Age'] >= 65, 'age_group'] = 'senior'
df.head(10)
```

```python
#2
total_passengers = len(df)
print(f"The total passengers in titanic are {total_passengers}")
```

```python
#3
age_count = df['age_group'].value_counts()
print(age_count)
```

```python
#4
proportions = age_count / total_passengers
print(proportions)
```

```python
#5
age_percent = (proportions) * 100
print(age_percent)
```
